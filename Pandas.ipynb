## Filter Operation

```python
import pandas as pd
data = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/student-data.json"
df = pd.read_json(data)
print(df)
```

```python
filter_data = df["maths"] > 50
# print(filter_data)
print(df[df["maths"] > 50])
```

```python
# english
print(df[df["english"] < 60])
```

```python
# physics marks equal to 78
print(df[df["physics"] == 78]) 
```

```python
# physics marks not-equal to 78
print(df[df["physics"] != 78]) 
```

```python
# name equal to sundar
print(df[df["name"] == "Sundar"])
```

```python
# print users where maths marks > 60 and english marks < 60
print(df[(df["maths"] > 60) & (df["english"] < 60)])
```

```python
# print users where gender is male and maths marks > 60
print((df[(df["gender"] == "Male") & (df["maths"] > 60)]))
```

```python
# print users where gender is male or maths marks > 60
print((df[(df["gender"] == "Male") | (df["maths"] > 60)]))
```

```python
# print physics > 60 and english < 60
print(df[(df["physics"] > 60) & (df["english"] < 60)]["physics"])
```

```python
print(df[(df["physics"] > 60) & (df["english"] < 60)][["physics","maths"]])
```
